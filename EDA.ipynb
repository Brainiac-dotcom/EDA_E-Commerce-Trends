# EXPLORATORY DATA ANALYSIS

```python
# Loading the dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv(r"C:\Users\DELL\Documents\EDA_E-Commerce Trends\Cleaned_Superstore.csv")
df
```

```python
# finding the statistics of all numeric values
df[["Sales","Quantity","Discount","Profit"]].describe()
```

```python
# Converting Order date & Ship date to data_time 

df["Ship Date"]= pd.to_datetime(df["Ship Date"])
df["Ship Date"]

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Order Date"]
```

```python
df.info()
```

```python
# Examine sales performane acroos categories and Regions
df.sample(5)
```

```python
# Which category has the highest sales
highest_sales_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
highest_sales_category
```

```python
# Calaculate the average sales per region 
Average_sales_per_region = df.groupby('Region')['Sales'].mean().sort_values(ascending=False)
Average_sales_per_region 
```

```python
# How does a particular category perform in a particular region
category_region_performance = df.pivot_table(index='Category', columns='Region', values='Sales', aggfunc='sum')
category_region_performance
```

```python
# The Average sales for each subcategory in each category
subcategories_per_category = df.pivot_table(index='Sub-Category', columns='Category', values='Sales', aggfunc='mean', fill_value=0)
subcategories_per_category
```

```python
# Calculate the sum  of sales of sub_category in category 
Sum_sales_sub_category_in_category = df.groupby(['Category', 'Sub-Category'])['Sales'].sum().reset_index()
Sum_sales_sub_category_in_category
```
